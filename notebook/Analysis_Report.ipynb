```python
import pandas as pd
df = pd.read_csv("data/data.csv", parse_dates=["date"])
df.head()
```

```python
df.groupby("city")["revenue"].sum().sort_values(ascending=False) #Total renvenue of every city
```

```python
df.groupby("station_id")["kWh_consumed"].mean()#power consumed by every stn
```

```python
df.groupby("peak_hours")["revenue"].mean()#revneue in peak vs n peak
```

```python
df["month"] = df["date"].dt.to_period("M")
df.groupby("month")["revenue"].sum() #monthly revenue 
```

```python
df["km_covered"] = df["kWh_consumed"] * 5
df["petrol_cost"] = (df["km_covered"] / 10) * 110
df["savings"] = df["petrol_cost"] - df["revenue"]
df.groupby("city")["savings"].mean()
```

```python
import matplotlib.pyplot as plt

# 1. Total Energy Consumption by City
city_energy = df.groupby("city")["kWh_consumed"].sum().sort_values(ascending=False)
plt.figure(figsize=(10,6))
city_energy.plot(kind="bar", color="skyblue")
plt.title("Total Energy Consumption by City")
plt.ylabel("kWh Consumed")
plt.xticks(rotation=45)
plt.show()
```

```python
city_revenue = df.groupby("city")["revenue"].sum().sort_values(ascending=False)
plt.figure(figsize=(10,6))
city_revenue.plot(kind="bar", color="orange")
plt.title("Total Revenue by City")
plt.ylabel("Revenue (₹)")
plt.xticks(rotation=45)
plt.show()
```

```python
peak_usage = df["peak_hours"].value_counts()
plt.figure(figsize=(6,6))
peak_usage.plot(kind="pie", autopct="%1.1f%%", labels=["Non-Peak","Peak"], colors=["lightgreen","red"])
plt.title("Peak vs Non-Peak Usage")
plt.ylabel("")
plt.show()
```

```python
avg_duration = df.groupby("station_id")["duration_mins"].mean().head(10)
plt.figure(figsize=(12,6))
avg_duration.plot(kind="bar", color="purple")
plt.title("Average Charging Duration (Top 10 Stations)")
plt.ylabel("Minutes")
plt.show()
```

```python
daily_energy = df.groupby("date")["kWh_consumed"].sum()
plt.figure(figsize=(14,6))
daily_energy.plot(color="blue")
plt.title("Daily Energy Consumption Trend")
plt.xlabel("Date")
plt.ylabel("kWh Consumed")
plt.show()
```
